## 5. Transform and Clean Data

```python
feature_cols   = [c for c in df.columns if c not in [TARGET]]
X_raw          = df[feature_cols].copy()
y_all          = df[TARGET].copy()
groups_all     = df[GROUP_COL].astype(str).copy() if GROUP_COL in df.columns else pd.Series(['ALL']*len(df))

numeric_cols    = X_raw.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols= [c for c in X_raw.columns if c not in numeric_cols]

num_pipe = Pipeline([('imputer', SimpleImputer(strategy='median'))])
cat_pipe = Pipeline([('imputer', SimpleImputer(strategy='most_frequent')),
                     ('onehot',  OneHotEncoder(handle_unknown='ignore'))])
prep = ColumnTransformer([
    ('num', num_pipe, numeric_cols),
    ('cat', cat_pipe, categorical_cols),
], remainder='drop')

# Geochemistry-only feature list (numeric, no lat/lon/state)
GEO_COLS = [c for c in numeric_cols if c not in ['Latitude','Longitude']]

print(f"Total features: {len(feature_cols)} | Numeric: {len(numeric_cols)} | Categorical: {len(categorical_cols)}")
print(f"Geochemistry columns: {GEO_COLS}")
X_raw.shape
```

## 6. Run Core Analysis

### T1.1 â€” Monotonic Constraints on XGBoost and LightGBM

```python
# Geothermal monotonic priors:
# SiO2 (+1), TDS (+1), Depth(ft) (+1), Na (+1), K (+1) => higher -> higher T
# pH (-1) => higher pH -> lower T for hydrothermal systems
# Constraints must map to the column order in the NUMERIC feature array

mono_map = {'SiO2': 1, 'TDS': 1, 'Depth(ft)': 1, 'Na': 1, 'K': 1, 'pH': -1}

t1_1_results = []

# --- XGBoost ---
if XGB_OK:
    # Build constraint tuple for numeric_cols order
    xgb_constraints = tuple(mono_map.get(c, 0) for c in numeric_cols)

    # Pipeline: median imputer only (XGBoost handles missings) + xgb
    xgb_pipe = Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('model',   xgb.XGBRegressor(
            n_estimators=300, max_depth=4, learning_rate=0.05,
            monotone_constraints=xgb_constraints,
            random_state=42, verbosity=0
        ))
    ])

    # Evaluate under random split (seed 0)
    idx = np.arange(len(X_raw[numeric_cols]))
    tr, te = train_test_split(idx, test_size=0.2, random_state=0)
    xgb_pipe.fit(X_raw[numeric_cols].iloc[tr], y_all.iloc[tr])
    pred_xgb = xgb_pipe.predict(X_raw[numeric_cols].iloc[te])
    rmse_xgb = float(np.sqrt(mean_squared_error(y_all.iloc[te], pred_xgb)))
    r2_xgb   = float(r2_score(y_all.iloc[te], pred_xgb))
    t1_1_results.append({'model':'xgboost_constrained','protocol':'random_split_seed0',
                          'rmse':rmse_xgb,'r2':r2_xgb,
                          'constraints_applied':str(mono_map)})
    print(f"XGBoost constrained -> RMSE={rmse_xgb:.2f}, R^2={r2_xgb:.3f}")

# --- LightGBM ---
if LGB_OK:
    lgb_constraints = [mono_map.get(c, 0) for c in numeric_cols]
    lgb_pipe = Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('model',   lgb.LGBMRegressor(
            n_estimators=300, max_depth=4, learning_rate=0.05,
            monotone_constraints=lgb_constraints,
            monotone_constraints_method='advanced',
            random_state=42, verbosity=-1
        ))
    ])
    lgb_pipe.fit(X_raw[numeric_cols].iloc[tr], y_all.iloc[tr])
    pred_lgb = lgb_pipe.predict(X_raw[numeric_cols].iloc[te])
    rmse_lgb = float(np.sqrt(mean_squared_error(y_all.iloc[te], pred_lgb)))
    r2_lgb   = float(r2_score(y_all.iloc[te], pred_lgb))
    t1_1_results.append({'model':'lgb_constrained','protocol':'random_split_seed0',
                          'rmse':rmse_lgb,'r2':r2_lgb,
                          'constraints_applied':str(mono_map)})
    print(f"LightGBM constrained -> RMSE={rmse_lgb:.2f}, R^2={r2_lgb:.3f}")

# Reference: unconstrained GBM
gbm_ref = Pipeline([('prep',prep),
                    ('model',GradientBoostingRegressor(n_estimators=300,max_depth=4,
                                                        learning_rate=0.05,random_state=42))])
gbm_ref.fit(X_raw.iloc[tr], y_all.iloc[tr])
pred_ref = gbm_ref.predict(X_raw.iloc[te])
rmse_ref = float(np.sqrt(mean_squared_error(y_all.iloc[te], pred_ref)))
r2_ref   = float(r2_score(y_all.iloc[te], pred_ref))
t1_1_results.append({'model':'gbm_unconstrained','protocol':'random_split_seed0',
                      'rmse':rmse_ref,'r2':r2_ref,'constraints_applied':'none'})
print(f"GBM unconstrained (reference) -> RMSE={rmse_ref:.2f}, R^2={r2_ref:.3f}")

t1_1_df = pd.DataFrame(t1_1_results)
t1_1_df
```

### T1.2 â€” SHAP TreeExplainer (Replace/Augment Permutation Importance)

```python
if not SHAP_OK:
    print("SHAP not available --- skipping T1.2")
else:
    # Train GBM on numeric features (SHAP TreeExplainer requires simple matrix input)
    gbm_shap = GradientBoostingRegressor(n_estimators=300, max_depth=4,
                                          learning_rate=0.05, random_state=42)
    idx      = np.arange(len(X_raw[numeric_cols]))
    tr, te   = train_test_split(idx, test_size=0.2, random_state=0)
    imp_num  = SimpleImputer(strategy='median')
    X_tr_num = imp_num.fit_transform(X_raw[numeric_cols].iloc[tr])
    X_te_num = imp_num.transform(X_raw[numeric_cols].iloc[te])
    gbm_shap.fit(X_tr_num, y_all.iloc[tr])

    explainer_shap = shap.TreeExplainer(gbm_shap)
    shap_vals      = explainer_shap.shap_values(X_te_num)

    # 1. Summary plot (beeswarm -- direction + magnitude)
    fig, ax = plt.subplots(figsize=(8, 5))
    shap.summary_plot(shap_vals, X_te_num, feature_names=numeric_cols, show=False)
    plt.tight_layout()
    fig.savefig(FIG_ROOT / 't1_2_shap_summary_beeswarm.png', dpi=160, bbox_inches='tight')
    plt.close('all')

    # 2. Bar plot (mean |SHAP|)
    fig, ax = plt.subplots(figsize=(8, 5))
    shap.summary_plot(shap_vals, X_te_num, feature_names=numeric_cols,
                      plot_type='bar', show=False)
    plt.tight_layout()
    fig.savefig(FIG_ROOT / 't1_2_shap_summary_bar.png', dpi=160, bbox_inches='tight')
    plt.close('all')

    # 3. Dependence plot for SiO2 (silica geothermometer proxy)
    if 'SiO2' in numeric_cols:
        sio2_idx = numeric_cols.index('SiO2')
        fig, ax = plt.subplots(figsize=(6, 4))
        shap.dependence_plot(sio2_idx, shap_vals, X_te_num,
                             feature_names=numeric_cols, ax=ax, show=False)
        ax.set_title('SHAP Dependence: SiO2 vs Temperature')
        fig.tight_layout()
        fig.savefig(FIG_ROOT / 't1_2_shap_dependence_SiO2.png', dpi=160, bbox_inches='tight')
        plt.close('all')

    # 4. Per-family mean |SHAP|
    cation_names = ['Na','K','Ca','Mg']
    anion_names  = ['Cl','SO4','HCO3','F','Br']
    proxy_names  = ['SiO2','TDS','pH','Depth(ft)']
    cation_idx   = [i for i,c in enumerate(numeric_cols) if c in cation_names]
    anion_idx    = [i for i,c in enumerate(numeric_cols) if c in anion_names]
    proxy_idx    = [i for i,c in enumerate(numeric_cols) if c in proxy_names]
    family_shap  = {
        'Cation':  float(np.abs(shap_vals[:, cation_idx]).mean()) if cation_idx else 0,
        'Anion':   float(np.abs(shap_vals[:, anion_idx]).mean())  if anion_idx  else 0,
        'Proxy':   float(np.abs(shap_vals[:, proxy_idx]).mean())  if proxy_idx  else 0,
    }
    mean_abs_shap = np.abs(shap_vals).mean(axis=0)
    shap_importance_df = pd.DataFrame({
        'feature':        numeric_cols,
        'mean_abs_shap':  mean_abs_shap,
    }).sort_values('mean_abs_shap', ascending=False)

    print("Family SHAP:", family_shap)
    print(shap_importance_df.head(10).to_string(index=False))
```

### T1.3 â€” KL Divergence to Quantify Distribution Shift (Random vs Spatial)

```python
from scipy.stats import entropy

def kl_divergence_binned(a, b, bins=30):
    """KL divergence between two 1-D arrays via histogram binning."""
    a, b = np.asarray(a, float), np.asarray(b, float)
    a, b = a[np.isfinite(a)], b[np.isfinite(b)]
    if len(a) < 2 or len(b) < 2:
        return np.nan
    lo, hi = min(a.min(), b.min()), max(a.max(), b.max())
    if lo == hi:
        return 0.0
    bins_e  = np.linspace(lo, hi, bins+1)
    pa, _   = np.histogram(a, bins=bins_e, density=True)
    pb, _   = np.histogram(b, bins=bins_e, density=True)
    pa = pa + 1e-10; pb = pb + 1e-10
    pa /= pa.sum();  pb /= pb.sum()
    return float(entropy(pa, pb))

# Random-split train/test (seed 0)
idx = np.arange(len(df))
tr_rand, te_rand = train_test_split(idx, test_size=0.2, random_state=0)

# Spatial: use largest held-out state as proxy
group_counts = groups_all.value_counts()
largest_state = group_counts.index[0]
te_spat = np.where(groups_all == largest_state)[0]
tr_spat = np.where(groups_all != largest_state)[0]

kl_rows = []
for col in GEO_COLS:
    if col not in df.columns:
        continue
    kl_rand  = kl_divergence_binned(df[col].iloc[tr_rand], df[col].iloc[te_rand])
    kl_spat  = kl_divergence_binned(df[col].iloc[tr_spat], df[col].iloc[te_spat])
    kl_rows.append({'feature': col, 'kl_random_split': kl_rand, 'kl_spatial_holdout': kl_spat})

kl_df = pd.DataFrame(kl_rows).sort_values('kl_spatial_holdout', ascending=False)
print(f"Largest KL under spatial holdout (state={largest_state}):")
print(kl_df.head(8).to_string(index=False))
```

```python
if not SHAP_OK:
    print("SHAP not available --- skipping T1.5")
else:
    # Model A: random split -- impute on train partition only
    tr_r, te_r = train_test_split(np.arange(len(X_raw[numeric_cols])), test_size=0.2, random_state=0)
    imp_r = SimpleImputer(strategy='median')
    X_tr_r = imp_r.fit_transform(X_raw[numeric_cols].iloc[tr_r])
    X_te_r = imp_r.transform(X_raw[numeric_cols].iloc[te_r])
    gbm_rand = GradientBoostingRegressor(n_estimators=200, max_depth=4,
                                          learning_rate=0.05, random_state=42)
    gbm_rand.fit(X_tr_r, y_all.iloc[tr_r])
    shap_rand = np.abs(shap.TreeExplainer(gbm_rand).shap_values(X_te_r)).mean(axis=0)

    # Model B: spatial holdout (largest state) -- impute on train partition only
    imp_s = SimpleImputer(strategy='median')
    X_tr_s = imp_s.fit_transform(X_raw[numeric_cols].iloc[tr_spat])
    X_te_s = imp_s.transform(X_raw[numeric_cols].iloc[te_spat])
    gbm_spat = GradientBoostingRegressor(n_estimators=200, max_depth=4,
                                          learning_rate=0.05, random_state=42)
    gbm_spat.fit(X_tr_s, y_all.iloc[tr_spat])
    shap_spat = np.abs(shap.TreeExplainer(gbm_spat).shap_values(X_te_s)).mean(axis=0)

    spear_r, spear_p = spearmanr(shap_rand, shap_spat)
    stability_df = pd.DataFrame({
        'feature':          numeric_cols,
        'shap_random':      shap_rand,
        'shap_spatial':     shap_spat,
        'rank_random':      pd.Series(shap_rand).rank(ascending=False).values,
        'rank_spatial':     pd.Series(shap_spat).rank(ascending=False).values,
    }).sort_values('shap_random', ascending=False)
    stability_df['rank_diff'] = (stability_df['rank_random'] - stability_df['rank_spatial']).abs()

    print(f"SHAP rank correlation (random vs spatial): r={spear_r:.3f}, p={spear_p:.4f}")
    if spear_r < 0.7:
        print("-> LOW stability: feature importance is NOT consistent across protocols")
    else:
        print("-> HIGH stability: feature importance is consistent across protocols")
    print(stability_df[['feature','rank_random','rank_spatial','rank_diff']].to_string(index=False))
```

### T2.7 â€” GMM Soft-Assignment Facies (Comparison to Hard Clustering)

```python
gmm_cols = GEO_COLS[:8] if len(GEO_COLS)>=8 else GEO_COLS
imp_gmm = SimpleImputer(strategy='median')
X_gmm = imp_gmm.fit_transform(X_raw[gmm_cols])
scaler_gmm = StandardScaler()
X_gmm_sc   = scaler_gmm.fit_transform(X_gmm)

gmm_results = []
for n_comp in [3, 4, 5, 6]:
    gmm = GaussianMixture(n_components=n_comp, covariance_type='full',
                           random_state=42, max_iter=200)
    gmm.fit(X_gmm_sc)
    labels     = gmm.predict(X_gmm_sc)
    probs      = gmm.predict_proba(X_gmm_sc)
    bic        = float(gmm.bic(X_gmm_sc))
    aic        = float(gmm.aic(X_gmm_sc))
    # Mean max-probability (how confident is the soft assignment?)
    mean_conf  = float(probs.max(axis=1).mean())
    # Temperature std within each component (lower = more homogeneous)
    temp_std_within = np.mean([float(y_all[labels == k].std()) for k in range(n_comp)
                               if (labels==k).sum() > 1])
    gmm_results.append({'n_components':n_comp,'BIC':bic,'AIC':aic,
                         'mean_confidence':mean_conf,'temp_std_within':temp_std_within})

gmm_df = pd.DataFrame(gmm_results)
best_n  = int(gmm_df.loc[gmm_df['BIC'].idxmin(), 'n_components'])
print(f"Best GMM by BIC: {best_n} components")
print(gmm_df.to_string(index=False))

# Soft-assignment entropy (how mixed are samples?)
best_gmm   = GaussianMixture(n_components=best_n, covariance_type='full', random_state=42)
best_gmm.fit(X_gmm_sc)
best_probs = best_gmm.predict_proba(X_gmm_sc)
entropy_gmm = -np.sum(best_probs * np.log(best_probs + 1e-10), axis=1)
print(f"Mean assignment entropy: {entropy_gmm.mean():.3f} (0=hard, log(k)={np.log(best_n):.2f}=max)")
```

### T4.1 - Charge Balance Error (CBE) Quality Filter

**Source:** Insights_DataCleaning_Workshop.md Â§6  
**Impact:** Removes analytically compromised samples (|CBE| > 15%); improves data quality  
**Papers:** Paper 1/2 methods section â€” document as pre-processing quality gate

```python
print("=== T4.1: Charge Balance Error (CBE) Quality Filter ===")

# CBE = (sum_cations - sum_anions) / (sum_cations + sum_anions)
# Cations: Na, K, Ca, Mg  (meq/L approx: divide by equivalent weight)
# Anions:  Cl, SO4, HCO3  (meq/L approx)
# Equivalent weights (g/eq): Na=23, K=39, Ca=20, Mg=12.15, Cl=35.45, SO4=48, HCO3=61

EQ = {'Na': 23.0, 'K': 39.1, 'Ca': 20.0, 'Mg': 12.15,
      'Cl': 35.45, 'SO4': 48.0, 'HCO3': 61.0}

df_cbe = df.copy()
cbe_rows = []
for col, ew in EQ.items():
    if col in df_cbe.columns:
        df_cbe[col + '_meq'] = df_cbe[col].fillna(0) / ew

cation_cols = [c + '_meq' for c in ['Na', 'K', 'Ca', 'Mg'] if c + '_meq' in df_cbe.columns]
anion_cols  = [c + '_meq' for c in ['Cl', 'SO4', 'HCO3'] if c + '_meq' in df_cbe.columns]

if cation_cols and anion_cols:
    sum_cat = df_cbe[cation_cols].sum(axis=1)
    sum_an  = df_cbe[anion_cols].sum(axis=1)
    denom   = (sum_cat + sum_an).replace(0, float('nan'))
    df_cbe['CBE_pct'] = 100.0 * (sum_cat - sum_an) / denom

    n_total    = len(df_cbe)
    n_bad      = (df_cbe['CBE_pct'].abs() > 15).sum()
    n_good     = n_total - n_bad
    pct_bad    = 100.0 * n_bad / n_total

    print(f"  Total samples     : {n_total}")
    print(f"  |CBE| <= 15 pct   : {n_good}  ({100-pct_bad:.1f}%)")
    print(f"  |CBE|  > 15 pct   : {n_bad}   ({pct_bad:.1f}%) -- potentially unreliable")
    print(f"  Mean CBE          : {df_cbe['CBE_pct'].mean():.2f}%")
    print(f"  Median CBE        : {df_cbe['CBE_pct'].median():.2f}%")

    # CBE as a soft feature (absolute value = quality signal)
    df['CBE_abs'] = df_cbe['CBE_pct'].abs()

    import pandas as pd_inner
    cbe_df = pd_inner.DataFrame({
        'metric': ['n_total', 'n_good_cbe', 'n_bad_cbe', 'pct_bad', 'mean_cbe', 'median_cbe'],
        'value':  [n_total, n_good, n_bad, round(pct_bad, 2),
                   round(df_cbe['CBE_pct'].mean(), 2), round(df_cbe['CBE_pct'].median(), 2)]
    })
    print("\n  CBE summary:")
    print(cbe_df.to_string(index=False))
else:
    print("  Not enough ion columns for CBE -- using placeholder")
    df['CBE_abs'] = 0.0
    cbe_df = pd.DataFrame({'metric': ['status'], 'value': ['insufficient_columns']})

print("\nT4.1 done.")
```

### T4.2 - Ionic Strength Proxy Feature

**Source:** Insights_DataCleaning_Workshop.md Â§6  
**Impact:** Captures overall salinity as a continuous feature; correlates with TDS but with physical basis  
**Formula:** I = 0.5 * sum(c_i * z_i^2) where z_i = ion charge

```python
print("=== T4.2: Ionic Strength Proxy Feature ===")
import numpy as np

# Ionic strength: I = 0.5 * sum(c_i * z_i^2) in meq/L units
# Charges: Na=1, K=1, Ca=2, Mg=2, Cl=1, SO4=2, HCO3=1
CHARGES = {'Na': 1, 'K': 1, 'Ca': 2, 'Mg': 2, 'Cl': 1, 'SO4': 2, 'HCO3': 1}
EQ2 = {'Na': 23.0, 'K': 39.1, 'Ca': 20.0, 'Mg': 12.15, 'Cl': 35.45, 'SO4': 48.0, 'HCO3': 61.0}

ionic_sum = pd.Series(0.0, index=df.index)
used_cols = []
for ion, z in CHARGES.items():
    if ion in df.columns:
        conc_meqL = df[ion].fillna(0) / EQ2[ion]
        ionic_sum += conc_meqL * (z ** 2)
        used_cols.append(ion)

df['ionic_strength'] = 0.5 * ionic_sum

print(f"  Ions used          : {used_cols}")
print(f"  ionic_strength min : {df['ionic_strength'].min():.4f}")
print(f"  ionic_strength max : {df['ionic_strength'].max():.4f}")
print(f"  ionic_strength mean: {df['ionic_strength'].mean():.4f}")

# Correlation with TDS and Temperature
corr_tds  = df['ionic_strength'].corr(df['TDS'])  if 'TDS'  in df.columns else float('nan')
corr_temp = df['ionic_strength'].corr(df[TARGET])

print(f"  Corr(ionic_str, TDS)        : {corr_tds:.3f}")
print(f"  Corr(ionic_str, Temperature): {corr_temp:.3f}")

ionic_df = pd.DataFrame({
    'feature': ['ionic_strength'],
    'min': [round(df['ionic_strength'].min(), 4)],
    'max': [round(df['ionic_strength'].max(), 4)],
    'mean': [round(df['ionic_strength'].mean(), 4)],
    'corr_with_TDS': [round(corr_tds, 3)],
    'corr_with_temp': [round(corr_temp, 3)]
})
print("\n", ionic_df.to_string(index=False))
print("\nT4.2 done.")
```

### T4.3 - Discrimination Threshold Tuning for Scaling Risk Classifier (Paper 4)

**Source:** Insights_ML_Project_Checklist.md Stage VII  
**Impact:** False negatives (missed high-risk wells) cost more than false positives -> optimal threshold != 0.5  
**Output:** Threshold sweep table showing precision/recall/F1 vs. threshold

```python
print("=== T4.3: Discrimination Threshold Tuning (Paper 4 Scaling Risk) ===")
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import train_test_split as tts
from sklearn.metrics import precision_score, recall_score, f1_score, balanced_accuracy_score
import numpy as np

# Build binary scaling-risk target (TDS > median = high risk)
if 'TDS' in df.columns:
    tds_med = df['TDS'].median()
    y_risk  = (df['TDS'] > tds_med).astype(int)
    feat_risk = [c for c in GEO_COLS if c != 'TDS'][:12]
    X_risk = df[feat_risk].copy()
    for c in feat_risk:
        X_risk[c] = X_risk[c].fillna(X_risk[c].median())

    X_tr, X_te, y_tr, y_te = tts(X_risk, y_risk, test_size=0.2, random_state=42, stratify=y_risk)
    clf_thresh = GradientBoostingClassifier(n_estimators=200, max_depth=3, random_state=42)
    clf_thresh.fit(X_tr, y_tr)
    y_prob = clf_thresh.predict_proba(X_te)[:, 1]

    thresholds = np.arange(0.10, 0.91, 0.05)
    rows = []
    for t in thresholds:
        y_pred = (y_prob >= t).astype(int)
        rows.append({
            'threshold': round(t, 2),
            'precision': round(precision_score(y_te, y_pred, zero_division=0), 3),
            'recall':    round(recall_score(y_te, y_pred, zero_division=0), 3),
            'f1':        round(f1_score(y_te, y_pred, zero_division=0), 3),
            'balanced_acc': round(balanced_accuracy_score(y_te, y_pred), 3)
        })

    thresh_df = pd.DataFrame(rows)
    best_f1_row = thresh_df.loc[thresh_df['f1'].idxmax()]
    best_ba_row = thresh_df.loc[thresh_df['balanced_acc'].idxmax()]

    print("  Threshold sweep (n_test={})".format(len(y_te)))
    print(thresh_df.to_string(index=False))
    print(f"\n  Best F1 threshold: {best_f1_row['threshold']} (F1={best_f1_row['f1']})")
    print(f"  Best balanced-acc threshold: {best_ba_row['threshold']} (BA={best_ba_row['balanced_acc']})")
    print("  Dissertation note: Use best_f1 threshold in Paper 4 -- false negatives (missed risk) cost more")
else:
    print("  TDS column not found -- skipping T4.3")
    thresh_df = pd.DataFrame({'status': ['TDS_not_found']})

print("\nT4.3 done.")
```

### T4.5 - Grouped Residuals by State

**Source:** Insights_04_06_Preprocessing_Feature_Engineering.md Chapter 4  
**Impact:** Shows where the model is weakest geographically; identifies state-level systematic bias  
**Output:** Per-state RMSE + mean residual table; flags states with worst performance

```python
print("=== T4.5: Grouped Residuals by State ===")
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import LeaveOneGroupOut
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_error
import numpy as np, warnings
warnings.filterwarnings('ignore')

X_gr = df[GEO_COLS].copy()
y_gr = df[TARGET].values
groups_gr = df[GROUP_COL].values

imp_gr = SimpleImputer(strategy='median')
X_gr_imp = imp_gr.fit_transform(X_gr)

logo = LeaveOneGroupOut()
state_rows = []

for train_idx, test_idx in logo.split(X_gr_imp, y_gr, groups=groups_gr):
    state_name = groups_gr[test_idx[0]]
    X_tr, X_te = X_gr_imp[train_idx], X_gr_imp[test_idx]
    y_tr, y_te = y_gr[train_idx], y_gr[test_idx]

    if len(y_te) < 3:
        continue

    m = GradientBoostingRegressor(n_estimators=150, max_depth=4, random_state=42)
    m.fit(X_tr, y_tr)
    y_pred = m.predict(X_te)
    residuals = y_te - y_pred

    state_rows.append({
        'state': state_name,
        'n_samples': len(y_te),
        'mean_temp': round(float(y_te.mean()), 2),
        'rmse': round(float(np.sqrt(mean_squared_error(y_te, y_pred))), 2),
        'mean_residual': round(float(residuals.mean()), 2),
        'abs_bias': round(float(np.abs(residuals.mean())), 2),
        'residual_std': round(float(residuals.std()), 2)
    })

residuals_by_state_df = pd.DataFrame(state_rows).sort_values('rmse', ascending=False)
print("  Per-state RMSE (LOSO):")
print(residuals_by_state_df.to_string(index=False))

worst3 = residuals_by_state_df.head(3)['state'].tolist()
best3  = residuals_by_state_df.tail(3)['state'].tolist()
print(f"\n  Worst 3 states (highest RMSE): {worst3}")
print(f"  Best  3 states (lowest  RMSE): {best3}")
print(f"  Overall mean RMSE across states: {residuals_by_state_df['rmse'].mean():.2f} degC")
print("  Dissertation note: Worst states may lack training-set analogues -> spatial data gap")

print("\nT4.5 done.")
```

**Impact:** Outlier-resistant scaling for SVR; better than StandardScaler when high-T outliers exist  
**Papers:** Paper 1 â€” SVR benchmark should use RobustScaler not StandardScaler

```python
print("=== N1.2: RobustScaler vs StandardScaler for SVR ===")
from sklearn.preprocessing import RobustScaler, StandardScaler
from sklearn.svm import SVR
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_error
import numpy as np

X_n2 = df[GEO_COLS].copy()
y_n2 = df[TARGET].values
from sklearn.model_selection import train_test_split as tts2
Xtr2, Xte2, ytr2, yte2 = tts2(X_n2, y_n2, test_size=0.2, random_state=42)

results_n12 = []
for scaler_name, scaler_cls in [('StandardScaler', StandardScaler), ('RobustScaler', RobustScaler)]:
    pipe = Pipeline([
        ('imp', SimpleImputer(strategy='median')),
        ('scaler', scaler_cls()),
        ('model', SVR(kernel='rbf', C=10, gamma='scale', epsilon=1.0))
    ])
    pipe.fit(Xtr2, ytr2)
    rmse = np.sqrt(mean_squared_error(yte2, pipe.predict(Xte2)))
    results_n12.append({'scaler': scaler_name, 'svr_rmse_degC': round(rmse, 2)})
    print(f"  SVR + {scaler_name}: RMSE = {rmse:.2f} degC")

n1_2_df = pd.DataFrame(results_n12)
# Check if RobustScaler improves for high-temperature outlier samples
high_T_mask = yte2 > np.percentile(yte2, 90)
print(f"\n  High-temperature samples (>90th pct): n={high_T_mask.sum()}")
print("  RobustScaler is preferred when spatial holdout reveals outlier states")
print("\nN1.2 done.")
```

### N1.3 - Learning Curves (Training Size vs. RMSE)

**Source:** ML Cookbook Ch.11 (DustinAlandzes)  
**Impact:** Shows whether more training data would close the protocol gap  
**Papers:** Paper 2 â€” justifies or refutes need for more geothermal data collection

```python
print("=== N1.3: Learning Curves (GroupKFold) ===")
from sklearn.model_selection import learning_curve, GroupKFold
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.impute import SimpleImputer
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

imp_lc = SimpleImputer(strategy='median')
X_lc = imp_lc.fit_transform(df[GEO_COLS])
y_lc = df[TARGET].values
groups_lc = df[GROUP_COL].values

train_sizes, train_scores, val_scores = learning_curve(
    GradientBoostingRegressor(n_estimators=150, max_depth=4, random_state=42),
    X_lc, y_lc,
    cv=GroupKFold(n_splits=5),
    groups=groups_lc,
    train_sizes=np.linspace(0.1, 1.0, 8),
    scoring='neg_root_mean_squared_error',
    n_jobs=-1
)

train_rmse = -train_scores.mean(axis=1)
val_rmse   = -val_scores.mean(axis=1)
train_std  = train_scores.std(axis=1)
val_std    = val_scores.std(axis=1)

# Plot
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(train_sizes, train_rmse, 'o-', color='steelblue', label='Training RMSE')
ax.fill_between(train_sizes, train_rmse - train_std, train_rmse + train_std, alpha=0.15, color='steelblue')
ax.plot(train_sizes, val_rmse, 'o-', color='darkorange', label='GroupKFold Val RMSE')
ax.fill_between(train_sizes, val_rmse - val_std, val_rmse + val_std, alpha=0.15, color='darkorange')
ax.set_xlabel('Training Set Size (samples)')
ax.set_ylabel('RMSE (degC)')
ax.set_title('Learning Curve (GroupKFold) â€” GBM on GGDB Geothermal')
ax.legend()
plt.tight_layout()
fig_path = FIG_ROOT / 'n1_3_learning_curve.png'
plt.savefig(fig_path, dpi=100, bbox_inches='tight')
plt.close()
print(f"  Figure: {fig_path}")

n1_3_df = pd.DataFrame({
    'train_size': train_sizes.astype(int),
    'train_rmse': train_rmse.round(2),
    'val_rmse':   val_rmse.round(2),
    'gap':        (val_rmse - train_rmse).round(2)
})
print(n1_3_df.to_string(index=False))

# Interpretation
still_descending = val_rmse[-1] < val_rmse[-2]
print(f"\n  Val RMSE still decreasing at max training size: {still_descending}")
if still_descending:
    print("  -> More data would likely improve performance (Paper 2 recommendation)")
else:
    print("  -> Performance has plateaued â€” data volume is not the bottleneck")
print("\nN1.3 done.")
```

### N1.4 - Gaussian Anomaly Detection (EllipticEnvelope)

**Source:** Homemade ML â€” Gaussian anomaly detection (ayoub-zmdumps)  
**Impact:** Flags geochemically anomalous samples for manual review without discarding them  
**Papers:** Paper 5 (Framework) + NB09 data curation quality gate

```python
print("=== N1.4: Gaussian Anomaly Detection (EllipticEnvelope) ===")
from sklearn.covariance import EllipticEnvelope
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
import numpy as np

imp_ae = SimpleImputer(strategy='median')
X_ae_raw = df[GEO_COLS].copy()
X_ae_imp = imp_ae.fit_transform(X_ae_raw)

scaler_ae = StandardScaler()
X_ae_scaled = scaler_ae.fit_transform(X_ae_imp)

# Fit on full dataset (unsupervised â€” no train/test split needed for flagging)
detector = EllipticEnvelope(contamination=0.05, random_state=42, support_fraction=0.9)
try:
    detector.fit(X_ae_scaled)
    anomaly_labels = detector.predict(X_ae_scaled)   # -1=anomaly, 1=normal
    anomaly_scores = detector.decision_function(X_ae_scaled)

    n_anomalies = (anomaly_labels == -1).sum()
    n_normal    = (anomaly_labels ==  1).sum()

    print(f"  Total samples  : {len(df)}")
    print(f"  Normal (flag=0): {n_normal} ({100*n_normal/len(df):.1f}%)")
    print(f"  Anomalous (flag=1): {n_anomalies} ({100*n_anomalies/len(df):.1f}%)")

    # Per-state anomaly rate
    df_ae = df[[GROUP_COL, TARGET]].copy()
    df_ae['anomaly_flag'] = (anomaly_labels == -1).astype(int)
    df_ae['anomaly_score'] = anomaly_scores

    state_anomaly = df_ae.groupby(GROUP_COL)['anomaly_flag'].agg(['sum','mean']).round(3)
    state_anomaly.columns = ['n_anomalies', 'anomaly_rate']
    state_anomaly = state_anomaly.sort_values('anomaly_rate', ascending=False)
    print("\n  Per-state anomaly rate:")
    print(state_anomaly.to_string())

    n1_4_df = pd.DataFrame({
        'metric': ['n_total', 'n_normal', 'n_anomalies', 'anomaly_pct', 'mean_score'],
        'value':  [len(df), n_normal, n_anomalies,
                   round(100*n_anomalies/len(df), 2),
                   round(anomaly_scores.mean(), 4)]
    })
    print("\n  Dissertation note: Flag anomalies for domain review, do NOT auto-remove")
except Exception as e:
    print(f"  EllipticEnvelope failed: {e}")
    n1_4_df = pd.DataFrame({'metric': ['status'], 'value': [str(e)[:60]]})
    df_ae = pd.DataFrame()

print("\nN1.4 done.")
```

### N1.6 - Train vs. Test Accuracy Gap (Overfitting Diagnostic)

**Source:** Nanodegree Decision Trees / Titanic (pdeguzman96)  
**Impact:** Explicit overfitting check for ALL classifiers; gap > 5% = overfitting flag  
**Papers:** Paper 4 (Scaling Risk) Table â€” report both train and test metrics

```python
print("=== N1.6: Train vs Test Accuracy Gap (Overfitting Diagnostic) ===")
from sklearn.ensemble import GradientBoostingClassifier, RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score
from sklearn.model_selection import train_test_split as tts6
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

if 'TDS' in df.columns:
    tds_med6 = df['TDS'].median()
    y_risk6  = (df['TDS'] > tds_med6).astype(int)
    feat6    = [c for c in GEO_COLS if c not in ['TDS']][:12]
    X6 = df[feat6].copy()
    imp6 = SimpleImputer(strategy='median')
    X6_imp = imp6.fit_transform(X6)
    scaler6 = StandardScaler()
    X6_sc = scaler6.fit_transform(X6_imp)

    Xtr6, Xte6, ytr6, yte6 = tts6(X6_sc, y_risk6, test_size=0.2, random_state=42, stratify=y_risk6)

    classifiers6 = {
        'DecisionTree(d=4)':  DecisionTreeClassifier(max_depth=4, random_state=42),
        'DecisionTree(d=8)':  DecisionTreeClassifier(max_depth=8, random_state=42),
        'RandomForest':       RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1),
        'GBC':                GradientBoostingClassifier(n_estimators=200, random_state=42),
        'LogisticRegression': LogisticRegression(max_iter=1000, random_state=42)
    }

    overfit_rows = []
    for name, clf in classifiers6.items():
        clf.fit(Xtr6, ytr6)
        tr_acc = accuracy_score(ytr6, clf.predict(Xtr6))
        te_acc = accuracy_score(yte6, clf.predict(Xte6))
        te_f1  = f1_score(yte6, clf.predict(Xte6))
        gap    = tr_acc - te_acc
        flag   = 'OVERFIT' if gap > 0.05 else 'OK'
        overfit_rows.append({'model': name, 'train_acc': round(tr_acc,3),
                              'test_acc': round(te_acc,3), 'gap': round(gap,3),
                              'test_f1': round(te_f1,3), 'overfit_flag': flag})
        print(f"  {name}: train={tr_acc:.3f}, test={te_acc:.3f}, gap={gap:.3f} [{flag}]")

    n1_6_df = pd.DataFrame(overfit_rows)
    n_overfit = (n1_6_df['overfit_flag'] == 'OVERFIT').sum()
    print(f"\n  {n_overfit}/{len(classifiers6)} classifiers flagged as overfitting (gap > 5%)")
    print("  Dissertation note: Report BOTH train and test metrics in Paper 4 Table")
else:
    print("  TDS not available -- skipping N1.6")
    n1_6_df = pd.DataFrame({'status': ['TDS_not_found']})

print("\nN1.6 done.")
```

### N1.7 - Regression Diagnostics Triptych

**Source:** Python DS & ML Bootcamp â€” linear regression diagnostics  
**Impact:** Standard 3-panel publication figure: residual histogram + predicted vs actual + QQ plot  
**Papers:** Paper 1 supplementary / Paper 2 diagnostics section

```python
print("=== N1.7: Regression Diagnostics Triptych ===")
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.model_selection import train_test_split as tts7
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_error, r2_score
import scipy.stats as stats_n17
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np

imp7 = SimpleImputer(strategy='median')
X7_imp = imp7.fit_transform(df[GEO_COLS])
y7 = df[TARGET].values

Xtr7, Xte7, ytr7, yte7 = tts7(X7_imp, y7, test_size=0.2, random_state=42)
gbm7 = GradientBoostingRegressor(n_estimators=200, max_depth=4, learning_rate=0.1, random_state=42)
gbm7.fit(Xtr7, ytr7)
ypred7 = gbm7.predict(Xte7)
resid7 = yte7 - ypred7

rmse7 = np.sqrt(mean_squared_error(yte7, ypred7))
r2_7  = r2_score(yte7, ypred7)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Panel 1: Residual histogram
axes[0].hist(resid7, bins=40, color='steelblue', alpha=0.75, edgecolor='white')
axes[0].axvline(0, color='red', linestyle='--', linewidth=1.5)
axes[0].axvline(resid7.mean(), color='orange', linestyle='--', linewidth=1.5, label=f'Mean={resid7.mean():.1f}')
axes[0].set_xlabel('Residual (degC)')
axes[0].set_ylabel('Count')
axes[0].set_title('Residual Distribution')
axes[0].legend(fontsize=9)

# Panel 2: Predicted vs Actual
axes[1].scatter(yte7, ypred7, alpha=0.35, s=12, color='steelblue')
lims = [min(yte7.min(), ypred7.min()), max(yte7.max(), ypred7.max())]
axes[1].plot(lims, lims, 'r--', linewidth=1.5, label='1:1 line')
axes[1].set_xlabel('Actual Temperature (degC)')
axes[1].set_ylabel('Predicted Temperature (degC)')
axes[1].set_title(f'Predicted vs Actual\nRMSE={rmse7:.2f} degC, R2={r2_7:.3f}')
axes[1].legend(fontsize=9)

# Panel 3: QQ plot
stats_n17.probplot(resid7, dist='norm', plot=axes[2])
axes[2].set_title('QQ Plot of Residuals\n(normality check)')

plt.suptitle('Regression Diagnostics â€” GBM on GGDB Geothermal (Random Split)', fontsize=12, y=1.02)
plt.tight_layout()
fig_path = FIG_ROOT / 'n1_7_regression_diagnostics.png'
plt.savefig(fig_path, dpi=100, bbox_inches='tight')
plt.close()
print(f"  Figure: {fig_path}")

# Normality test on residuals
_, p_shapiro = stats_n17.shapiro(resid7[:min(5000, len(resid7))])
skew7 = float(pd.Series(resid7).skew())
kurt7 = float(pd.Series(resid7).kurt())
print(f"  Residual mean   : {resid7.mean():.3f} degC  (near 0 = unbiased)")
print(f"  Residual std    : {resid7.std():.2f} degC")
print(f"  Skewness        : {skew7:.3f}  (|<0.5| = approximately normal)")
print(f"  Kurtosis        : {kurt7:.3f}")
print(f"  Shapiro-Wilk p  : {p_shapiro:.4f}  ({'normal' if p_shapiro > 0.05 else 'non-normal'})")

n1_7_df = pd.DataFrame({
    'metric': ['rmse', 'r2', 'resid_mean', 'resid_std', 'skewness', 'kurtosis', 'shapiro_p'],
    'value':  [round(rmse7,2), round(r2_7,3), round(float(resid7.mean()),3),
               round(float(resid7.std()),2), round(skew7,3), round(kurt7,3), round(p_shapiro,4)]
})
print("\nN1.7 done.")
```

```python
# N10.5 + N4.1 â€” Multi-class ROC-AUC + F-Beta (beta=2) for regime classification
# For Papers 7/8 if geothermal regime (low/medium/high) is modeled as classification
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import cross_validate, GroupKFold
from sklearn.metrics import make_scorer, roc_auc_score, fbeta_score

_data_path = Path(r'c:/Users/obalo/Downloads/PhDOneDrive/PhD_Research_Operating_System/03_data_processed/geothermal_canonical_v2_0_0.csv')
_df = pd.read_csv(_data_path)
_TARGET = 'Temperature(C)'; _GROUP = 'State'

# 3-class regime: low (<100C), medium (100-200C), high (>200C)
_df['regime'] = pd.cut(_df[_TARGET], bins=[-np.inf, 100, 200, np.inf],
                        labels=['low', 'medium', 'high'])
_y_cls = _df['regime']
_X = _df.drop(columns=[_TARGET, 'regime']); _g = _df[_GROUP]

_num_cols = _X.select_dtypes(include=[np.number]).columns.tolist()
_cat_cols = [c for c in _X.columns if c not in _num_cols]
_prep = ColumnTransformer([
    ('num', SimpleImputer(strategy='median'), _num_cols),
    ('cat', Pipeline([('imp', SimpleImputer(strategy='most_frequent')),
                      ('ohe', OneHotEncoder(handle_unknown='ignore'))]), _cat_cols),
])
_rf_cls = Pipeline([('prep', _prep),
    ('clf', RandomForestClassifier(n_estimators=300, random_state=42, n_jobs=-1))])

_roc_scorer = make_scorer(roc_auc_score, needs_proba=True, multi_class='ovr', average='weighted')
_f2_scorer = make_scorer(fbeta_score, beta=2, average='weighted')

_cv = cross_validate(_rf_cls, _X, _y_cls, groups=_g, cv=GroupKFold(n_splits=5),
    scoring={'roc_auc_ovr': _roc_scorer, 'f2': _f2_scorer})
print("=== N10.5+N4.1 Regime Classification (3-class) ===")
print(f"CV ROC-AUC (OvR, weighted): {_cv['test_roc_auc_ovr'].mean():.4f} +/- {_cv['test_roc_auc_ovr'].std():.4f}")
print(f"CV F2 (weighted):           {_cv['test_f2'].mean():.4f} +/- {_cv['test_f2'].std():.4f}")
print("F2 (beta=2) penalizes missed high-temperature wells by weighting recall.")
```
